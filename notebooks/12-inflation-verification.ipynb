# Module 12: Inflation-Linked Bonds

## TIPS Pricing & Real Yields

```
from src.inflation.curves import generate_real_curve
from src.inflation.tips import price_tips
import numpy as np

# Real vs Nominal
nominal_yields = np.array([0.04, 0.045, 0.05])
breakeven_inflation = np.array([0.02, 0.022, 0.025])
real_yields = generate_real_curve(nominal_yields, breakeven_inflation)

print("Nominal Yields:", nominal_yields)
print("Breakeven Inflation:", breakeven_inflation)
print("Implied Real Yields:", real_yields)

# TIPS Pricing
cpi_base = 250.0
cpi_current = 265.0
index_ratio = cpi_current / cpi_base

# Price of a TIPS bond
face_value = 1000
coupon_rate = 0.015  # Real coupon
maturity = 5
real_ytm = 0.01

price = price_tips(face_value, coupon_rate, maturity, real_ytm, index_ratio)
print(
    f"\nTIPS Price (Real YTM {real_ytm:.2%}, Index Ratio {index_ratio:.4f}): ${price:.2f}"
)
```
