# Module 12: Inflation-Linked Bonds (TIPS)

This notebook verifies the implementation of the Inflation-Linked Bonds (TIPS) module.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.inflation.tips import TIPS
from src.inflation.indexing import daily_ref_cpi, index_ratio
from src.inflation.curves import bootstrap_real_rates, breakeven_inflation

# 1. TIPS Initialization
tips = TIPS(face_value=100.0, coupon_rate=0.02, maturity=5.0, freq=2, base_cpi=250.0)
print(f"TIPS initialized: {tips}")
```

```python
# 2. Inflation Adjusted Cashflows
index_ratios = np.linspace(1.0, 1.1, tips.periods) # simulating 2% annual inflation
t, cf = tips.inflation_adjusted_cashflows(index_ratios)
plt.bar(t, cf)
plt.title('Inflation-Adjusted Cashflows')
plt.xlabel('Period')
plt.ylabel('Cashflow')
plt.show()
```

```python
# 3. Breakeven Inflation
maturities = np.array([1, 2, 3, 5, 10])
nominal_spots = np.array([0.04, 0.041, 0.042, 0.045, 0.05])
real_spots = np.array([0.01, 0.011, 0.013, 0.015, 0.018])

breakevens = breakeven_inflation(nominal_spots, real_spots)
plt.plot(maturities, nominal_spots, label='Nominal Spot')
plt.plot(maturities, real_spots, label='Real Spot')
plt.plot(maturities, breakevens, label='Breakeven Inflation', linestyle='--')
plt.legend()
plt.title('Breakeven Inflation Curve')
plt.xlabel('Maturity (Years)')
plt.ylabel('Rate')
plt.show()
```
