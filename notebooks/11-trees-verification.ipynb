## Binomial Interest Rate Tree (Module 45)
This section demonstrates the recombining log-normal binomial tree for short rate modeling and zero-coupon bond pricing via backward induction.

```python
from src.trees.binomial import BinomialTree

# Create a 5-step binomial tree
volatility = 0.20  # 20% volatility
time_step = 1.0  # 1-year step
r0 = 0.05  # 5% initial short rate

tree = BinomialTree(r0=r0, volatility=volatility, time_step=time_step, periods=5)

# Calculate theoretical ZCB price for a face value of $1.0
zcb_price = tree.price_zero_coupon_bond(face_value=1.0)
print(f"Zero-Coupon Bond Price (5-year, r0=5%): ${zcb_price * 100:.2f}")

# Display rates at step 2
rates = tree.build_tree()
rates_step_2 = rates[2]
print(f"Rates at Step 2: {[f'{r * 100:.2f}%' for r in rates_step_2]}")
```
